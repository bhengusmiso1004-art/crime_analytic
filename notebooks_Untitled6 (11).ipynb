```python
import re

def clean_station_name(name):
    name = str(name).lower().strip()

    # Replace punctuation with spaces
    name = re.sub(r"[^a-z0-9\s]", " ", name)

    # Remove extra spaces
    name = re.sub(r"\s+", " ", name).strip()

    return name
```

```python
from sklearn.model_selection import train_test_split

# Features
X = ml_df[
    [
        "Province",
        "Year",
        "population",
        "area_km2"
    ]
].copy()

# Target
y = ml_df["Hotspot"].copy()

# Split: 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

categorical_features = [
    "Province",
    "Year"
]

numeric_features = [
    "population",
    "area_km2"
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "num",
            StandardScaler(),
            numeric_features
        )
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

import pandas as pd

# ---------------------------------
# Models to compare
# ---------------------------------
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=5
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        random_state=42
    )
}

# Store results and trained pipelines
results = []
trained_models = {}

# ---------------------------------
# Train and evaluate every model
# ---------------------------------
for name, model in models.items():

    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("classifier", model)
        ]
    )

    # Train
    pipeline.fit(X_train, y_train)

    # Predict
    y_pred = pipeline.predict(X_test)

    # Calculate metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(
        y_test, y_pred, zero_division=0
    )
    recall = recall_score(
        y_test, y_pred, zero_division=0
    )
    f1 = f1_score(
        y_test, y_pred, zero_division=0
    )

    # Save results
    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1_Score": f1
    })

    # Save trained model
    trained_models[name] = pipeline


# ---------------------------------
# Comparison table
# ---------------------------------
results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="F1_Score",
    ascending=False
).reset_index(drop=True)

results_df
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

for name, model in trained_models.items():

    y_pred = model.predict(X_test)

    ConfusionMatrixDisplay.from_predictions(
        y_test,
        y_pred
    )

    plt.title(f"Confusion Matrix - {name}")
    plt.show()
```

```python
from sklearn.metrics import classification_report

for name, model in trained_models.items():

    print("=" * 60)
    print(name)
    print("=" * 60)

    y_pred = model.predict(X_test)

    print(
        classification_report(
            y_test,
            y_pred,
            target_names=["Non-Hotspot", "Hotspot"]
        )
    )
```

```python
from sklearn.metrics import accuracy_score, f1_score

decision_tree = trained_models["Decision Tree"]

# Predictions
train_pred = decision_tree.predict(X_train)
test_pred = decision_tree.predict(X_test)

print("DECISION TREE PERFORMANCE")
print("-" * 40)

print("Training Accuracy:",
      round(accuracy_score(y_train, train_pred), 4))

print("Testing Accuracy:",
      round(accuracy_score(y_test, test_pred), 4))

print("Training F1:",
      round(f1_score(y_train, train_pred), 4))

print("Testing F1:",
      round(f1_score(y_test, test_pred), 4))
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline

# Pipeline
dt_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", DecisionTreeClassifier(random_state=42))
])

# Parameters to test
param_grid = {
    "classifier__max_depth": [3, 5, 7, 10, 15, 20],
    "classifier__min_samples_split": [2, 5, 10, 20],
    "classifier__min_samples_leaf": [1, 2, 5, 10]
}

# Grid search
grid_search = GridSearchCV(
    estimator=dt_pipeline,
    param_grid=param_grid,
    scoring="f1",
    cv=5,
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

print("Best parameters:")
print(grid_search.best_params_)

print("\nBest Cross-Validation F1:")
print(round(grid_search.best_score_, 4))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

# Best model found by GridSearchCV
tuned_dt = grid_search.best_estimator_

# Predictions
train_pred_tuned = tuned_dt.predict(X_train)
test_pred_tuned = tuned_dt.predict(X_test)

# Training results
print("TUNED DECISION TREE")
print("=" * 45)

print("\nTraining Accuracy:",
      round(accuracy_score(y_train, train_pred_tuned), 4))

print("Training F1:",
      round(f1_score(y_train, train_pred_tuned), 4))

# Test results
print("\nTesting Accuracy:",
      round(accuracy_score(y_test, test_pred_tuned), 4))

print("Testing Precision:",
      round(precision_score(y_test, test_pred_tuned), 4))

print("Testing Recall:",
      round(recall_score(y_test, test_pred_tuned), 4))

print("Testing F1:",
      round(f1_score(y_test, test_pred_tuned), 4))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        test_pred_tuned,
        target_names=["Non-Hotspot", "Hotspot"]
    )
)
```

```python
# ------------------------------------------
# TIME-BASED TRAIN / TEST SPLIT
# ------------------------------------------

train_years = [
    "2013-2014",
    "2014-2015",
    "2015-2016",
    "2016-2017",
    "2017-2018",
    "2018-2019",
    "2019-2020",
    "2020-2021"
]

test_years = [
    "2021-2022",
    "2022-2023"
]

# Training data
time_train = ml_df[
    ml_df["Year"].isin(train_years)
].copy()

# Testing data
time_test = ml_df[
    ml_df["Year"].isin(test_years)
].copy()

print("Training years:")
print(sorted(time_train["Year"].unique()))

print("\nTesting years:")
print(sorted(time_test["Year"].unique()))

print("\nTraining observations:", len(time_train))
print("Testing observations:", len(time_test))

print("\nTraining hotspot distribution:")
print(time_train["Hotspot"].value_counts())

print("\nTesting hotspot distribution:")
print(time_test["Hotspot"].value_counts())
```

```python
# Convert year to a numeric value
time_train["Year_Number"] = (
    time_train["Year"].str[:4].astype(int)
)

time_test["Year_Number"] = (
    time_test["Year"].str[:4].astype(int)
)

# Features
feature_columns = [
    "Province",
    "Year_Number",
    "population",
    "area_km2"
]

X_train_time = time_train[feature_columns]
y_train_time = time_train["Hotspot"]

X_test_time = time_test[feature_columns]
y_test_time = time_test["Hotspot"]

print("Training X:", X_train_time.shape)
print("Testing X:", X_test_time.shape)

print("\nTraining years:")
print(sorted(time_train["Year_Number"].unique()))

print("\nTesting years:")
print(sorted(time_test["Year_Number"].unique()))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

import pandas as pd


# Preprocessing
time_preprocessor = ColumnTransformer([
    (
        "cat",
        OneHotEncoder(handle_unknown="ignore"),
        ["Province"]
    ),
    (
        "num",
        StandardScaler(),
        ["Year_Number", "population", "area_km2"]
    )
])


# Models
time_models = {

    "Tuned Decision Tree":
        DecisionTreeClassifier(
            max_depth=20,
            min_samples_leaf=2,
            min_samples_split=10,
            random_state=42
        ),

    "Random Forest":
        RandomForestClassifier(
            n_estimators=200,
            random_state=42,
            n_jobs=-1
        ),

    "Gradient Boosting":
        GradientBoostingClassifier(
            random_state=42
        )
}


time_results = []
time_trained_models = {}


for name, classifier in time_models.items():

    model = Pipeline([
        ("preprocessor", time_preprocessor),
        ("classifier", classifier)
    ])

    model.fit(X_train_time, y_train_time)

    predictions = model.predict(X_test_time)

    time_results.append({
        "Model": name,
        "Accuracy": accuracy_score(
            y_test_time, predictions
        ),
        "Precision": precision_score(
            y_test_time, predictions,
            zero_division=0
        ),
        "Recall": recall_score(
            y_test_time, predictions,
            zero_division=0
        ),
        "F1_Score": f1_score(
            y_test_time, predictions,
            zero_division=0
        )
    })

    time_trained_models[name] = model


time_results_df = pd.DataFrame(time_results)

time_results_df = time_results_df.sort_values(
    "F1_Score",
    ascending=False
).reset_index(drop=True)

time_results_df
```

```python
from sklearn.metrics import ConfusionMatrixDisplay, classification_report
import matplotlib.pyplot as plt

# Final classification model
final_model = time_trained_models["Random Forest"]

# Final predictions
final_predictions = final_model.predict(X_test_time)

# Confusion matrix
ConfusionMatrixDisplay.from_predictions(
    y_test_time,
    final_predictions,
    display_labels=["Non-Hotspot", "Hotspot"]
)

plt.title("Random Forest - Hotspot Classification")
plt.show()

# Final classification report
print(
    classification_report(
        y_test_time,
        final_predictions,
        target_names=["Non-Hotspot", "Hotspot"]
    )
)
```
